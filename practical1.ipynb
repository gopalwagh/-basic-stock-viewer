# EXPERIMENT 1: Introduction to python

# Operator

```python
a = 10
b = 3

# Arithmetic Operators
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)

# Comparison Operators
print(a > b)
print(a == b)
print(a != b)

# Assignment Operator
a += 5
print(a)
```

# Data types

```python
a = 10
b = 10.5
c = "Hello"
d = True
e = [1, 2, 3]
f = (1, 2, 3)
g = {1, 2, 3}
h = {"name": "Rahul", "age": 20}

print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
print(type(f))
print(type(g))
print(type(h))
```

# Logical Operator

```python
a = 10
b = 5

print(a > 5 and b < 10)
print(a < 5 or b < 10)
print(not(a > 5))
```

# If- else

```python
marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")
```

#List

```python
fruits = ["apple", "banana", "mango"]

print(fruits[0])

fruits.append("orange")
fruits.remove("banana")

print(fruits)
```

# Dictionary

```python
student = {
    "name": "Rahul",
    "age": 20,
    "marks": 85
}

print(student["name"])

student["age"] = 21
student["city"] = "Nanded"

print(student)
```

#SET

```python
numbers = {1, 2, 3, 2, 1}

print(numbers)

numbers.add(4)
numbers.remove(2)

print(numbers)
```
